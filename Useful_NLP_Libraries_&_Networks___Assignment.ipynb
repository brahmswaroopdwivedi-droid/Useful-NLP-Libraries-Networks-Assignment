2.  What is TextBlob and how does it simplify common NLP tasks like

Built on top of NLTK and Pattern, it streamlines common tasks like sentiment analysis, translation, and text preprocessing (tokenization, POS tagging) into simple, one-line commands.
How TextBlob Simplifies NLP Tasks
TextBlob treats textual data as Python string-like objects, allowing for rapid prototyping and easy implementation of complex NLP pipelines.
- Sentiment Analysis: TextBlob provides a .sentiment property that returns polarity and subjectivity instantly.
1. Polarity: Ranges from -1.0 (negative) to 1.0 (positive).
2. Subjectivity: Ranges from 0.0 (objective) to 1.0 (subjective).
3. Code: TextBlob("I love this!").sentiment.
- Translation and Detection: It supports seamless language detection and translation using Google Translate's API, making multilingual processing trivial.
1. Code: blob.translate(to='es') (Translates to Spanish)
- Simplified Preprocessing: It makes tokenization, part-of-speech tagging, and noun phrase extraction as simple as calling properties like .words or .tags.
- Spelling Correction: TextBlob includes a .correct() method to fix spelling mistakes automatically.

4. Describe the architecture and functioning of a Recurrent Natural Network
(RNN).
- Recurrent Neural Networks (RNNs) are a type of deep learning model designed for sequential data, such as text or time series, featuring hidden state loops that store past information. Unlike traditional feedforward networks, RNNs share weights across time steps, allowing them to process variable-length data by updating their hidden state with the current input and previous hidden state.

5. What is the key difference between LSTM and GRU networks in NLP
applications?
- The key difference between LSTM and GRU in NLP is that LSTMs use three gates (input, forget, output) with a separate memory cell, making them better for complex, long-term dependencies. In contrast, GRUs use two gates (update, reset) to merge states, offering faster training and higher efficiency for smaller datasets or simpler tasks.

```python
#6. Write a Python program using TextBlob to perform sentiment analysis on the following paragraph of text: “I had a great experience using the new mobile banking app. The interface is intuitive, and customer support was quick to resolve my issue. However, the app did crash once
#during a transaction, which was frustrating" Your program should print out the polarity and subjectivity scores.
# First, install textblob if you haven't: pip install textblob
from textblob import TextBlob

# The input paragraph
text = "I had a great experience using the new mobile banking app. The interface is intuitive, and customer support was quick to resolve my issue. However, the app did crash once #during a transaction, which was frustrating"

# Create a TextBlob object
blob = TextBlob(text)

# Perform sentiment analysis
# polarity ranges from -1.0 (negative) to 1.0 (positive)
# subjectivity ranges from 0.0 (objective) to 1.0 (subjective)
sentiment = blob.sentiment

# Print the results
print(f"Text: {text}\n")
print(f"Polarity: {sentiment.polarity}")
print(f"Subjectivity: {sentiment.subjectivity}")

```

```python
#7. Given the sample paragraph below, perform string tokenization and frequency distribution using Python and NLTK: “Natural Language Processing (NLP) is a fascinating field that combines linguistics,computer science, and artificial intelligence. It enables machines to understand,
#interpret, and generate human language. Applications of NLP include chatbots,sentiment analysis, and machine translation. As technology advances, the role of NLP in modern solutions is becoming increasingly critical.”
import nltk
from nltk.tokenize import word_tokenize
from nltk.probability import FreqDist

# Download the required resource for tokenization
nltk.download('punkt')

sample_paragraph = """Natural Language Processing (NLP) is a fascinating field that combines linguistics,
computer science, and artificial intelligence. It enables machines to understand, #interpret,
and generate human language. Applications of NLP include chatbots,sentiment analysis,
and machine translation. As technology advances, the role of NLP in modern solutions
is becoming increasingly critical."""

# 1. String Tokenization
# Breaks the text into individual words and punctuation
tokens = word_tokenize(sample_paragraph)
print("Tokens:", tokens)

# 2. Frequency Distribution
# Calculates how often each word or punctuation mark appears
freq_dist = FreqDist(tokens)

# Display the 5 most frequent items
print("\nTop 5 Most Frequent Tokens:")
for word, frequency in freq_dist.most_common(5):
    print(f"{word}: {frequency}")
```

```python
#8. Implement a basic LSTM model in Keras for a text classification task using the following dummy dataset. Your model should classify sentences as either positive (1) or negative (0).
# Dataset texts = [
#“I love this project”, #Positive
#“This is an amazing experience”, #Positive
#“I hate waiting in line”, #Negative
#“This is the worst service”, #Negative
#“Absolutely fantastic!” #Positive]
#labels = [1, 1, 0, 0, 1]
#Preprocess the text, tokenize it, pad sequences, and build an LSTM model to train on this data. You may use Keras with TensorFlow backend.
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

# 1. Dataset
texts = [
    "I love this project",                # Positive
    "This is an amazing experience",      # Positive
    "I hate waiting in line",             # Negative
    "This is the worst service",          # Negative
    "Absolutely fantastic!"               # Positive
]
labels = np.array([1, 1, 0, 0, 1])

# 2. Preprocessing
vocab_size = 1000
max_length = 5
embedding_dim = 16

# Tokenize text
tokenizer = Tokenizer(num_words=vocab_size, oov_token="<OOV>")
tokenizer.fit_on_texts(texts)
sequences = tokenizer.texts_to_sequences(texts)

# Pad sequences
padded_sequences = pad_sequences(sequences, maxlen=max_length, padding='post', truncating='post')

# 3. Build Model
model = Sequential([
    Embedding(vocab_size, embedding_dim, input_length=max_length),
    LSTM(32),
    Dense(1, activation='sigmoid') # Binary classification
])

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# 4. Train Model
model.fit(padded_sequences, labels, epochs=10, verbose=0)

# 5. Evaluate/Predict
print("Model trained successfully.")
```

```python
#9. Using spaCy, build a simple NLP pipeline that includes tokenization,lemmatization, and entity recognition. Use the following paragraph as your dataset:“Homi Jehangir Bhaba was an Indian nuclear physicist who played a key role in the development of India’s atomic energy program. He was the founding director of the Tata
#Institute of Fundamental Research (TIFR) and was instrumental in establishing the Atomic Energy Commission of India.”Write a Python program that processes this text using spaCy, then prints tokens, their lemmas, and any named entities found.
import spacy

# 1. Load the English NLP model
# This model includes tokenization, lemmatization, and NER capabilities.
nlp = spacy.load("en_core_web_sm")

# 2. The dataset
text = """Homi Jehangir Bhaba was an Indian nuclear physicist who played a key role
in the development of India’s atomic energy program. He was the founding director
of the Tata Institute of Fundamental Research (TIFR) and was instrumental in
establishing the Atomic Energy Commission of India."""

# 3. Process the text
# This acts as the pipeline, performing tokenization, POS tagging, and parsing.
doc = nlp(text)

# 4. Print Tokens and their Lemmas
print(f"{'Token':<20} | {'Lemma':<20}")
print("-" * 43)
for token in doc:
    # token.text gives the token, token.lemma_ gives the base form
    print(f"{token.text:<20} | {token.lemma_:<20}")

# 5. Named Entity Recognition (NER)
print("\nNamed Entities Found:")
print("-" * 43)
for ent in doc.ents:
    # ent.text is the entity string, ent.label_ is the type
    print(f"{ent.text:<20} | {ent.label_:<20}")

```

2. Proposed Architecture
A Sequence-to-Sequence (Seq2Seq) encoder-decoder model is ideal for generating empathetic, human-like responses.
-Encoder (Input Analysis): An LSTM or GRU layer processes the embedded user input one token at a time. LSTMs are preferred for long-term dependencies, while GRUs are often faster for real-time applications.
- Hidden State (Context): The encoder compresses the input into a "context vector" that represents the user's current intent or emotional state.
- Decoder (Response Generation): A second recurrent layer takes this context vector and predicts the response sequence word-by-word.
- Attention Mechanism: This allows the model to "focus" on specific parts of a user's sentence (like the word "hopeless") when generating a response.
